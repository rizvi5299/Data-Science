````python
import pandas as pd
from scipy import stats
````

````python
df = pd.DataFrame(
    {
        "Headline A": [14, 986],
        "Headline B": [8, 992],
        "Headline C": [12, 988],
    },
    index=["Click", "No-click"]
)

df
````

````python
# to do a significance test, first create the null world under null hypothesis assumption

row_totals = df.sum(axis=1)
col_totals = df.sum(axis=0)
grand_total = df.values.sum()
````

````python
expected = pd.DataFrame(
    {
        col: [
            row_totals["Click"] * col_totals[col] / grand_total,
            row_totals["No-click"] * col_totals[col] / grand_total,
        ]
        for col in df.columns
    },
    index=df.index
)
````

````python
# this is if we assume the null that they're all the same
expected
````

````python
# create table of Pearson Residuals
r = (df - expected)/(expected**0.5)
r
````

````python
# chi-squared stat (X) is the total sum

chi_stat = (r**2).sum().sum()
chi_stat
````

````python
# degrees of freedom
df_chi = (df.shape[0] - 1) * (df.shape[1] - 1)
````

````python
# p value
p_value = 1 - stats.chi2.cdf(chi_stat, df_chi)
p_value
````

p-value is much greater than 0.05, so strong evidence to support the null. The difference of the click rates of the three headlines is not statistically significant. 

````python
# another test
# null states that voting tendencies is the same for both genders.
votes = pd.DataFrame(
    ((200,150,50),
    (250, 300, 50)),
    columns=['Rep', 'Dem', 'Ind'],
    index=['Male', 'Female']
)

votes
````

````python
chisq, pvalue, df, expected = stats.chi2_contingency(votes)
````

````python
expected
````

````python
chisq, pvalue
````

The p-value is much less than 0.05, so the probability of these results being by random chance is very low. 

Statistically, we can say that the differences between their results is not by random chance and that we have strong evidence to reject the null.

So voting tendencies do differ by sex, according to the p-value calculated from the samples.
