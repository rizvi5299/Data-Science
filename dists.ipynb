```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

# `scipy.stats` Distribution Methods

In `scipy.stats`, each distribution (e.g., `norm`, `poisson`, `binom`, `weibull_min`) has a consistent set of methods for probability calculations, statistics, sampling, and parameter estimation.

<h4> 1. Probability Functions </h4>
- <strong>Continuous distributions:</strong>  


  - `pdf(x, ...)` – Probability density function at `x`.  


  - `logpdf(x, ...)` – Natural logarithm of the pdf.  
- <strong>Discrete distributions:</strong>


  - `pmf(k, ...)` – Probability mass function at `k`.  



  - `logpmf(k, ...)` – Logarithm of the pmf.

<h4> 2. Cumulative Functions </h4>


- `cdf(x, ...)` – Cumulative distribution function (P(X ≤ x)).  


- `sf(x, ...)` – Survival function (1 − CDF, P(X > x)).  


- `logcdf(x, ...)` / `logsf(x, ...)` – Log versions.  


- `ppf(q, ...)` – Percent point function, inverse of `cdf` (quantiles).  


- `isf(q, ...)` – Inverse survival function, inverse of `sf`.

<h4> 3. Statistics and Moments </h4>


- `rvs(size=..., ...)` – Generate random samples.  


- `mean(...)` – Mean of the distribution. Same as .moment(1)


- `var(...)` – Variance. 


- `std(...)` – Standard deviation.  


- `moment(n, ...)` – n-th non-central moment.  


- `median(...)` – Median.  


- `entropy(...)` – Differential entropy.

<h4> 4. Parameter Estimation </h4>


- `fit(data, ...)` – Fit distribution parameters to data.  


- `freeze(**params)` – Create a frozen distribution with fixed parameters, so you can call `pdf`, `cdf`, etc., without specifying parameters each time.


```python
stats.binom.pmf(5, n=10, p=0.5) 
```

```python
stats.poisson.pmf(k=1, mu=2)
```

```python

nums = stats.norm.rvs(size=1000, loc=0, scale=1)
plt.hist(nums)
plt.show()
```

```python
nums.mean(), nums.std()
```

```python
poisson = stats.poisson.rvs(mu=3, size=10000)
plt.hist(poisson)
plt.show()
```

# Graphing a probability distribution


Use stem plots for discrete

```python
k_range = np.arange(25) # x axis

dist_pmf = stats.poisson.pmf(k_range, mu=10) # y axis


plt.stem(k_range, dist_pmf)
plt.xlabel('Number of events (k)')
plt.ylabel('Probability P(X=k)')
plt.title('Poisson Distribution (λ=3)')
plt.show()
```

```python
normal_k = np.arange(21)
normal_pdf = stats.norm.pdf(normal_k, loc=10, scale=1)

plt.plot(normal_k, normal_pdf)
plt.title("Normal distribution, std=1")
plt.show()
```
